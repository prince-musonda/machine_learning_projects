# In this simple network, we try to discover the  simple law of motion that the relationship between velocity and position with respect to time:



$x_{i+1}$ = $x_t$ + $v$ $\Delta\text{T}$

where $\Delta\text{T}$ = 1, for this simple network

```python
import numpy as np
import torch
```

## Dataset initialization

```python
rng = np.random.default_rng(42)
torch.manual_seed(42)

current_positions = rng.random.uniform(-10,10,1000)
velocities = rng.random.uniform(-5,5,1000)

next_positions = current_positions + velocities
```

```python
x = np.column_stack((current_positions,velocities))
```

```python
x = torch.tensor(x,dtype=torch.float32)
y  = torch.tensor(next_positions, dtype = torch.float32)
```

```python
y = y.unsqueeze(1)
```

## Model initialization (basic linear layer)

```python
model = torch.nn.Linear(2,1)
optim = torch.optim.SGD(model.parameters())
criterion = torch.nn.MSELoss()
```

## Model Training

```python
for epoch in range(800):
  prediction = model(x)
  loss= criterion(prediction,y)
  optim.zero_grad()
  loss.backward()
  optim.step()

  if epoch%10 == 0:
    print(f"epoch: {epoch},      loss: {loss:.6f}")
```

```python
print(model.weight)
```

```python
print(model.bias)
```

# model evaluation

```python
test_rng = np.random.default_rng(123)

# generate new test positions and velocities
test_positions1 = test_rng.uniform(-10,10,200)
test_positions2 = test_rng.uniform(100,200,200)
test_velocities = test_rng.uniform(-5,5,200)

# targets (correct answers)
next_positions1 = test_positions1 + test_velocities
next_positions2 = test_positions2 + test_velocities

# stack  positions with velocities
test1_input = np.column_stack((test_positions1, test_velocities))
test2_input = np.column_stack((test_positions2, test_velocities))

# convert to pytorch tensors
test1_input = torch.tensor(test1_input,dtype=torch.float16)
test2_input = torch.tensor(test2_input,dtype=torch.float16)

test1_target = torch.tensor(next_positions1, dtype=torch.float16).unsqueeze(1)
test2_target = torch.tensor(next_positions2, dtype=torch.float16).unsqueeze(1)
```

```python
model.eval()
model.to(dtype=torch.float16)
with torch.no_grad():
  predictions1 = model(test1_input)
  loss = criterion(predictions1, test1_target)
  print(f"first test loss: {loss.item():}")

  prediction2 = model(test2_input)
  loss = criterion(prediction2, test2_target)
  print(f"seconf test loss: {loss.item()}")
```

```python
# model weights:
print(model.weight)
print(model.bias)
```
